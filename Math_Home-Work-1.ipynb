## Задача 1. Робота з зображеннями як з матрицями

M = [[100, 150, 200], [50, 100, 150], [0, 50, 100]]
E = [[20, 30, 40], [10, 20, 30], [5, 10, 15]]

1) Зміна контрасту: 0.5 * M
= [[50, 75, 100], [25, 50, 75], [0, 25, 50]]

2) Корекція яскравості: M + 25
= [[125, 175, 225], [75, 125, 175], [25, 75, 125]]

3) Змішування: 0.8*M + 0.2*E
= [[84, 126, 168], [42, 84, 126], [1, 42, 83]]


```python
import numpy as np

M = np.array([
    [100, 150, 200],
    [50, 100, 150],
    [0, 50, 100]
], dtype=float)

E = np.array([
    [20, 30, 40],
    [10, 20, 30],
    [5, 10, 15]
], dtype=float)

contrast = 0.5 * M
brightness = M + 25
blend = 0.8 * M + 0.2 * E

print("0.5 * M:")
print(contrast)
print()
print("M + 25:")
print(brightness)
print()
print("0.8*M + 0.2*E:")
print(blend)
```


## Задача 2. Пошук схожих фільмів

u = (8, 2, 5)
v_A = (9, 1, 2)
v_B = (1, 9, 8)
v_C = (7, 2, 6)

Формула косинусної подібності:
cos(theta) = (u · v) / (||u|| * ||v||)

Норма профілю користувача:
||u|| = sqrt(8^2 + 2^2 + 5^2) = sqrt(93) ≈ 9.6437

Скалярні добутки:
u·v_A = 84
u·v_B = 66
u·v_C = 90

Косинусні подібності (наближено):
sim(u, v_A) ≈ 0.9393
sim(u, v_B) ≈ 0.5664
sim(u, v_C) ≈ 0.9892

Найкращий фільм: C (Drama Movie).


```python
import numpy as np

u = np.array([8, 2, 5], dtype=float)
movies = {
    "A (Action Movie)": np.array([9, 1, 2], dtype=float),
    "B (Comedy Movie)": np.array([1, 9, 8], dtype=float),
    "C (Drama Movie)": np.array([7, 2, 6], dtype=float),
}

def cosine_similarity(a, b):
    return np.dot(a, b) / (np.linalg.norm(a) * np.linalg.norm(b))

print("||u|| =", np.linalg.norm(u))

scores = {name: cosine_similarity(u, vec) for name, vec in movies.items()}
for name, score in scores.items():
    print(f"{name}: {score:.6f}")

best_movie = max(scores, key=scores.get)
print()
print("Найбільш схожий фільм:", best_movie)
```


## Задача 3. Аналіз виробництва

Матриця коефіцієнтів (стовпці: x1 Розвідник, x2 Камікадзе, x3 Вантажний):
A = [[4, 4, 6], [1, 1, 2], [1, 2, 4]]

Вектор витрат:
b = [460, 130, 240]

det(A) = -2, тобто не 0 => система має єдиний розв'язок.

Розв'язок:
x = [20, 50, 30]

Перевірка: A @ x = b.


```python

import numpy as np

A = np.array([
    [4, 4, 6],
    [1, 1, 2],
    [1, 2, 4]
], dtype=float)

b = np.array([460, 130, 240], dtype=float)

det_A = np.linalg.det(A)
print("det(A) =", det_A)

if np.isclose(det_A, 0):
    print("Система не має єдиного розв'язку")
else:
    x = np.linalg.solve(A, b)
    print("Розв'язок x =", x)

    check = A @ x
    print("Перевірка A @ x =", check)
    print("Чи збігається з b:", np.allclose(check, b))
```


## Задача 4. Прогноз тренду

t = [1, 2, 3, 4, 5]
y = [22, 28, 37, 45, 53]

Будуємо модель y = k*t + b методом найменших квадратів.

Через np.linalg.lstsq отримуємо:
k = 7.9, b = 13.3

Прогноз на 6-ту годину:
y_hat = 7.9*6 + 13.3 = 60.7

Бонус (нормальні рівняння):
Розв'язуємо (A^T A)x = A^T y, де x = [k, b].
Коефіцієнти збігаються з lstsq.


```python
import numpy as np

t = np.array([1, 2, 3, 4, 5], dtype=float)
y = np.array([22, 28, 37, 45, 53], dtype=float)

A = np.column_stack([t, np.ones_like(t)])

(k, b), residuals, rank, s = np.linalg.lstsq(A, y, rcond=None)
y_pred_6 = k * 6 + b

print(f"Рівняння тренду: y = {k:.4f} * t + {b:.4f}")
print(f"Прогноз на 6-ту годину: {y_pred_6:.4f}")

# Бонус: ручна реалізація через нормальні рівняння
ATA = A.T @ A
ATy = A.T @ y
x_normal = np.linalg.solve(ATA, ATy)

print()
print("Коефіцієнти через нормальні рівняння [k, b]:", x_normal)
print("Збіг з lstsq:", np.allclose(x_normal, [k, b]))
```
